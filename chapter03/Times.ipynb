```python
FN = 'Darius-13-100m-fly.txt'
FOLDER = 'swimdata/'
```

```python
swimmer, age, distance, stroke = FN.removesuffix('.txt').split('-')
```

```python
print(swimmer)
print(age)
print(distance)
print(stroke)
```

```python
with open(FOLDER + FN) as f:
    lines = f.readlines()
    times = lines[0].strip().split(',')
```

```python
converts = []
for time in times:
    minutes, rest = time.split(':')
    seconds, hundreds = rest.split('.')
    converted_time = (int(minutes) * 100 * 60) + (int(seconds) * 100) + int(hundreds)
    converts.append(converted_time)
    
converts
```

```python
import statistics as stat

average = stat.mean(converts)

mins_secs, hundredths = str(round(average / 100, 2)).split('.')
```

```python
minutes = int(mins_secs) // 60

seconds = int(mins_secs) - minutes * 60
```

```python
print(minutes)
print(seconds)
print(hundredths)
```

```python
average = str(minutes) + ":" + str(seconds) + '.' + str(hundredths)
```
